#### **7. Perform memory information on the data and recommend what non-default data types would you recommend to optimize memory settings for the dataframe.**

```python
print(df.nunique())

#Print out the different data types in the DataFrame
print(df.info(memory_usage="deep=True"))

#create a copy of the DataFrame to work with and call it df_downsized
df_downsized = df.copy()

# Downcast the integer columns to the smallest possible integer type to save memory
integer_columns = df_downsized.select_dtypes(include=["int"]).columns
df_downsized[integer_columns] = df_downsized[integer_columns].apply(pd.to_numeric, downcast="integer")

#downcast the object columns to the smallest possible object type to save memory
object_columns = df_downsized.select_dtypes(include=["object"]).columns
df_downsized[object_columns] = df_downsized[object_columns].apply(lambda x: x.astype("category") if x.nunique() / len(x) < 0.5 else x)

print(df_downsized.info(memory_usage="deep=True"))


# print out the memory usage of each dataframe but do not break it down by column
print(f"\n\nOriginal DataFrame memory usage: {df.memory_usage(index=False, deep=True).sum()/1024**2:.2f} MB")
print(f"Downsized DataFrame memory usage: {df_downsized.memory_usage(index=False, deep=True).sum()/1024**2:.2f} MB")
```

##### **MEMORY ANALYSIS**
After analyzing the dataframe, changing the default datatypes appears to be the biggest way to reduce memory storage. Columns that are **int64** should be reduced to int8 or int16 data type because the data contained in those columns are not large numbers requiring int64. 

Similarly, the columns that are of **object** data type should be changed to category data type. 

These 2 changes reduce the dataframe size significantly. The original size was **2.43MB**, but with the suggested data type changes, the size drops to **0.16MB** 

Those columns that are float data type should not be touched and remain float so as to keep their precision
